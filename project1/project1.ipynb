````python
from openai import OpenAI

# Configuration
openai_api_key = "YOU CAN PLACE YOUR KEY HERE"

# or you can load your key from a file to keep it private
key_file = "my_key.txt"

try:
    with open(key_file, 'r') as f:
        openai_api_key = f.read()
except FileNotFoundError:
    print(f"Error: The file '{key_file}' was not found.")
except Exception as e:
    print(f"An error occurred: {e}")

openai_api_base = "https://llm-api.arc.vt.edu/api/v1"

# Set up the OpenAI client
client = OpenAI(
	base_url = openai_api_base,
    api_key = openai_api_key,
)

model = "gpt-oss-120b"
````

````python
def converse(prompt, messages=None, model=model, max_tokens=5000):
    # Add the user's message to the list of messages
    if messages is None:
        messages = []

    messages.append({"role": "user", "content": prompt})

    response = client.chat.completions.create(
                messages=messages,
                model=model,
                max_tokens=max_tokens).choices[0].message.content

    # Add the assistant's message to the list of messages
    messages.append({"role": "assistant", "content": response})

    return response, messages
````

````python
messages = []
prompt=f"""
Write a formal, precise requirements statement in English for a python program
that takes an integer as input, representing a balance to make change for.

The purpose of the program is to work out how many 100, 50, 20, 10, 5 and 1
dollar notes are required to add up to the provided total.
"""
response, messages = converse(prompt, messages)
print(f"You: {prompt}\nChatGPT: {response}")

````
