# Overview

In this project, we are going to use the class LLM to help us become more familiar with the Python programming language.

This is an example of how we can use AI to change the process of software engineering: by asking it to help us learn new things.


```python
from openai import OpenAI

# Configuration
openai_api_key = "YOU CAN PLACE YOUR KEY HERE"

# Or you can load your key from a file to keep it private
key_file = "my_key.txt"

try:
    with open(key_file, 'r') as f:
        openai_api_key = f.read()
except FileNotFoundError:
    print(f"Error: The file '{key_file}' was not found.")
except Exception as e:
    print(f"An error occurred: {e}")

openai_api_base = "https://llm-api.arc.vt.edu/api/v1"

# Set up the OpenAI client
client = OpenAI(
	base_url = openai_api_base,
    api_key = openai_api_key,
)

model = "gpt-oss-120b"

def get_completion(prompt, model=model):
	messages = [{"role": "user", "content": prompt}]

	response = client.chat.completions.create(
                messages=messages,
                model=model,
                max_tokens=1000,
		        temperature=0)  # this is the degree of randomness of the model's output

	return response.choices[0].message.content
```
